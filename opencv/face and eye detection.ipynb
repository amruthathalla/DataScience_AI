```python
import cv2

# Use the full path to the XML if it's not in your script's folder
face_classifier = cv2.CascadeClassifier('Haarcascades/haarcascade_frontalface_default.xml')

image_path = r"C:\Users\Amrutha Thalla\FSDS\DataScience_AI\opencv\images (3).jpg"
image = cv2.imread(image_path)
print(image)

if image is None:
    print(f"Error: Could not load image at {image_path}")
else:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # Detect faces
    faces = face_classifier.detectMultiScale(gray, 1.3, 5)

    if len(faces) == 0:
     print("No faces found")
    else:
    # Draw rectangles on all detected faces
   
     for (x,y,w,h) in faces:
        cv2.rectangle(image, (x,y), (x+w,y+h), (127,0,255), 2)
    
    # Show the final result once
    cv2.imshow('Face Detection', image)
    cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python
import cv2

face_classifier = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
eye_classifier = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_eye.xml')

img = cv2.imread(r'C:\Users\Amrutha Thalla\FSDS\DataScience_AI\opencv\images (3).jpg')

if img is None:
    print("Error: Could not load image")
else:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    faces = face_classifier.detectMultiScale(gray, 1.3, 5)

    for (x, y, w, h) in faces:
        cv2.rectangle(img, (x, y), (x + w, y + h), (127, 0, 255), 2)

        roi_gray = gray[y:y + h, x:x + w]
        roi_color = img[y:y + h, x:x + w]

        roi_gray = cv2.equalizeHist(roi_gray)

        eyes = eye_classifier.detectMultiScale(
            roi_gray,
            scaleFactor=1.03,
            minNeighbors=3,
            minSize=(10, 10)
        )

        print("Eyes detected:", len(eyes))

        for (ex, ey, ew, eh) in eyes:
            cv2.rectangle(roi_color, (ex, ey), (ex + ew, ey + eh), (0, 255, 255), 3)

    # Show both windows
    cv2.imshow('Face & Eye Detection', img)
    cv2.imshow('ROI', roi_gray)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
